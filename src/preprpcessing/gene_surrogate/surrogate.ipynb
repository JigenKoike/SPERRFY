```python
import numpy as np
import time
from datetime import datetime, timedelta
from brainsmash.mapgen.base import Base
from scipy.io import savemat

num_genes = 763
num_surrogates = 1000
num_regions = 213
dist_mat_file = "distance_matrix.txt"

surrogate_genes = np.zeros((num_surrogates, num_regions, num_genes), dtype=np.float32)

start_time = time.time()

for i in range(1, num_genes + 1):
    gene_file = f"gene{i}.txt"
    try:
        base = Base(
            x=gene_file,
            D=dist_mat_file,
            resample=True
        )
        surrogates = base(n=num_surrogates)
        surrogate_genes[:, :, i - 1] = surrogates
    except Exception as e:
        print(f"Error at gene {i} ({gene_file}): {e}")
        continue

savemat("surrogate_genes_org.mat", {"surrogate_genes": surrogate_genes})
print(f"Saved successfully at {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
```
